```python
import random
import numpy as np
import matplotlib.pyplot as plt

true_rewards = [2, 5, 1, 8]
num_actions = len(true_rewards)
steps = 1000


#different epsilon values
epsilons = [0.0, 0.1, 0.3, 1.0]

total_rewards = []

for epsilon in epsilons:
  Q = [0] * num_actions
  N = [0] * num_actions
  total_reward = 0

  for step in range(steps):

    #exploration
    if random.random() < epsilon:
      action = random.randint(0, num_actions - 1)

    #exploitation
    else:
      action = np.argmax(Q)

    reward = np.random.normal(true_rewards[action], 1)
    total_reward += 1

    N[action] += 1
    Q[action] = Q[action] + (reward - Q[action]) / N[action]

  total_rewards.append(total_reward)


#plot
plt.figure(figsize=(7,5))
plt.bar([str(e) for e in epsilons], total_rewards)

plt.xlabel("Epsilon (ε)")
plt.ylabel("Total Reward")
plt.title("Comparison of Exploration and Exploitation")
plt.grid(axis='y')

plt.show()
```

```python
import random
import numpy as np
import matplotlib.pyplot as plt

true_rewards = [2, 5, 1, 8]

num_actions = len(true_rewards)
epsilon = 0.1
steps = 1000

Q = [0] * num_actions
N = [0] * num_actions

for step in range(steps):

    if random.random() < epsilon:
        action = random.randint(0, num_actions - 1)
    else:
        action = np.argmax(Q)

    reward = np.random.normal(true_rewards[action],1)

    N[action] += 1
    Q[action] += (reward - Q[action]) / N[action]

plt.figure(figsize=(7,5))
plt.bar(range(1,5), N)

plt.xlabel("Machine Number")
plt.ylabel("Times Selected")
plt.title("Number of Times Each Machine Was Selected")

plt.xticks([1,2,3,4])

plt.show()
```
